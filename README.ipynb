
# Employee Management MIS Application (Oracle PL/SQL)

This project is a comprehensive **Employee Management MIS Application** built using **Oracle PL/SQL**. It demonstrates core concepts such as database design, joins, stored procedures, cursors, transactions, packages, batch/bulk processing, PL/SQL blocks, and use of data dictionary views.

---

##  Project Structure

| Script File                  | Description                                                                 |
|-----------------------------|-----------------------------------------------------------------------------|
| `01_create_tables.sql`      | Creates all required tables for Employee Payroll database                   |
| `02_insert_sample_data.sql` | Inserts 15 sample rows into each table                                      |
| `03_joins.sql`              | Demonstrates INNER, LEFT, RIGHT, and FULL joins between employee-related tables |
| `04_batches_packages_bulk.sql` | Implements payroll generation using batch and bulk processing in packages   |
| `05_blocks.sql`             | Demonstrates usage of Anonymous and Named PL/SQL Blocks                     |
| `06_data_dictionary.sql`    | Uses Oracle data dictionary views to analyze and manage the schema          |

---

##  Features Demonstrated

###  Tables & Sample Data
- Employees, Departments, Salaries, Attendance, Payroll
- Logical foreign key relationships and constraints
- Sample data for realistic scenarios

###  SQL Joins
- Inner Join, Left Join, Right Join, Full Outer Join
- Applied to employee-department-attendance-payroll reporting

###  PL/SQL Procedures & Cursors
- Stored procedures using implicit and explicit cursors
- Dynamic reporting based on attendance status
- Transaction handling with `COMMIT` and `ROLLBACK`

###  PL/SQL Packages & Bulk Processing
- Package with payroll calculation logic
- `BULK COLLECT` and `FORALL` for high-performance processing

###  PL/SQL Blocks
- Anonymous block for single employee calculation
- Named procedure for reusability and modularity

###  Data Dictionary Queries
- Inspect tables, columns, indexes, constraints, dependencies
- Monitor object status and source code

## ▶ How to Run

1. Open SQL*Plus, Oracle SQL Developer, or any Oracle-compatible tool.
2. Run scripts in this order:

```
@01_create_tables.sql
@02_insert_sample_data.sql
@03_joins.sql
@04_batches_packages_bulk.sql
@05_blocks.sql
@06_data_dictionary.sql
```

>  Make sure the user has `DBMS_OUTPUT` enabled and proper privileges.

---

##  Requirements

- Oracle 19c or higher
- A user/schema with privileges to create tables, procedures, and packages
- Enable `DBMS_OUTPUT` in your SQL client
